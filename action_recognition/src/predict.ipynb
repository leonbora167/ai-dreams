# Simplified Inference Notebook

This notebook runs video inference with overlay output using `test.py`.

Assumption: your conda env is already activated.

```python
from pathlib import Path
import os
from IPython.display import Video

# Run from project root even if notebook opens inside src/
project_root = Path.cwd().resolve().parent if Path.cwd().name == "src" else Path.cwd().resolve()
os.chdir(project_root)
print("project_root:", project_root)
```

```python
# Pick latest checkpoint automatically (best.pt preferred)
runs_dir = project_root / "runs"
run_dirs = sorted([p for p in runs_dir.glob("*") if p.is_dir()], key=lambda p: p.stat().st_mtime, reverse=True)
assert run_dirs, "No runs found. Train first."
latest = run_dirs[0]
checkpoint = latest / "best.pt"
if not checkpoint.exists():
    checkpoint = latest / "last.pt"
assert checkpoint.exists(), f"No checkpoint found in {latest}"
print("using checkpoint:", checkpoint)
```

```python
# Set input/output video paths
input_video = project_root / "data_training" / "val" / "violent" / "cam1__96.mp4"
output_video = project_root / "runs" / "inference_output.mp4"
print("input:", input_video)
print("output:", output_video)
```

```python
# Run inference (tqdm progress bars appear in output)
!python test.py --video "$input_video" --checkpoint "$checkpoint" --output "$output_video" --config configs/train_example.yaml --stride 5
```
